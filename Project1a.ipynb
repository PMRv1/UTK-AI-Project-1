print("Size:", baseball_df.size)
This shows the total number of cells in the dataset (rows × columns). It helps understand the overall dataset size.

print("Shape (rows, cols):", baseball_df.shape)
This prints the number of rows and columns. It tells us how many players and features are in the dataset.

print("\nBasic Statistics:")
This just adds a label for clarity in the output.

print(baseball_df.describe())
This shows basic statistics like mean, min, max, and standard deviation for all numeric columns. It helps us quickly understand the distribution of numeric data.

print("\nMissing Values Count:")
This adds a label to indicate the next output is about missing data.

print(baseball_df.isnull().sum())
This counts how many missing values are in each column. It helps identify which columns need cleaning before analysis.

plt.figure(figsize=(10,6))
This creates a blank figure with a width of 10 inches and height of 6 inches for the plot. It sets the size of the histogram we are about to make.

plt.hist(baseball_df['Salary'].dropna(), bins=20, edgecolor='black', alpha=0.8)
This draws a histogram of player salaries, ignoring missing values. It divides salaries into 20 bins and adds black edges to each bar for clarity.

plt.title('Histogram of Player Salaries')
This adds a title to the histogram so it is clear what the chart represents.

plt.xlabel('Salary ($)')
This labels the X-axis as Salary in dollars.

plt.ylabel('Frequency')
This labels the Y-axis as the number of players in each salary range.

plt.grid(axis='y', alpha=0.75)
This adds horizontal grid lines to make it easier to read the histogram.

plt.tight_layout()
This adjusts the layout so that labels and titles fit nicely and don’t overlap.

plt.show()
This displays the histogram in the notebook so we can visualize the salary distribution.

```python
# Inspect dataset
print("Size:", baseball_df.size)
print("Shape (rows, cols):", baseball_df.shape)
print("\nBasic Statistics:")
print(baseball_df.describe())
print("\nMissing Values Count:")
print(baseball_df.isnull().sum())

# Histogram of Salary
plt.figure(figsize=(10,6))
plt.hist(baseball_df['Salary'].dropna(), bins=20, edgecolor='black', alpha=0.8)
plt.title('Histogram of Player Salaries')
plt.xlabel('Salary ($)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()

# Box plot of Salary by League
plt.figure(figsize=(10,6))
sns.boxplot(x='League', y='Salary', data=baseball_df)
plt.title('Box Plot of Salary by League')
plt.xlabel('League')
plt.ylabel('Salary ($)')
plt.show()

# Correlation heatmap of numeric features
plt.figure(figsize=(10,8))
correlation = baseball_df.select_dtypes(include=[np.number]).corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
